# **📊 Data Visualization**

# Load Cleaned Dataset

```python

import pandas as pd

df = pd.read_csv('/content/paper_leaks_cleaned.csv')

pd.set_option('display.max_rows', None)

display(df)

```

# Create Year Column

```python
# Create Year Column
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['year'] = df['date'].dt.year

print("Year column created successfully")
display(df[['date', 'year']])
```

# 1. How did paper leak cases change over the years?

```python
# Paper Leak Cases by Year
import matplotlib.pyplot as plt

yearly_leaks = df.groupby('year').size()

plt.figure(figsize=(10,5))
plt.plot(yearly_leaks.index, yearly_leaks.values, marker='o')
plt.xlabel('Year')
plt.ylabel('Number of Paper Leak Cases')
plt.title('Paper Leak Cases by Year')
plt.grid(True)
plt.show()
```

# 2. Which areas reported the most paper leak cases?

```python
# Top 10 Areas with Most Paper Leak Cases

import matplotlib.pyplot as plt

top_areas = df['area'].value_counts().head(10).sort_values()

plt.figure(figsize=(10, 6))
plt.barh(top_areas.index, top_areas.values)

plt.xlabel('Number of Paper Leak Cases')
plt.ylabel('Area')
plt.title('Top 10 Areas with Most Paper Leak Cases')

plt.tight_layout()
plt.show()
```

# 3. Which exams had the most paper leak cases?

```python
# Top 10 Exams with Most Paper Leak Cases

import matplotlib.pyplot as plt

top_exams = df['exam_name'].value_counts().head(10).sort_values()

plt.figure(figsize=(10, 6))
plt.barh(top_exams.index, top_exams.values)

plt.xlabel('Number of Paper Leak Cases')
plt.ylabel('Exam')
plt.title('Top 10 Exams with Most Paper Leak Cases')

plt.tight_layout()
plt.show()
```

# 4. How are paper leak cases distributed by leak status?

```python
# Paper Leak Cases by Leak Status

import matplotlib.pyplot as plt

status_counts = df['leak_status'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    status_counts.values,
    labels=status_counts.index,
    autopct='%1.1f%%'
)

plt.title('Distribution of Paper Leak Cases by Status')
plt.show()
```

# 5. How do UPA and NDA eras compare in number of cases?

```python
# Paper Leak Cases by Era

import matplotlib.pyplot as plt

era_counts = df['era'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(era_counts.index, era_counts.values)

plt.xlabel('Era')
plt.ylabel('Number of Paper Leak Cases')
plt.title('Paper Leak Cases by Era')

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# 6. Is there a relationship between arrests and convictions?

```python
# Relationship Between Arrests and Convictions

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(df['arrests'], df['convictions'])

plt.xlabel('Arrests')
plt.ylabel('Convictions')
plt.title('Relationship Between Arrests and Convictions')

plt.tight_layout()
plt.show()
```

# **📌 Results & Interpretation**

# 1. Which year recorded the highest number of paper leak cases?

```python
# Highest Paper Leak Year

year_counts = df.groupby('year').size()

highest_year = year_counts.idxmax()
highest_cases = year_counts.max()
print("Shows the year with the most paper leak cases.")
print("Year with the highest paper leak cases:", highest_year)
print("Number of cases:", highest_cases)
```

# 2. Which area reported the most paper leak cases?

```python
# Area with the Most Paper Leak Cases

area_counts = df['area'].value_counts()

highest_area = area_counts.idxmax()
highest_area_cases = area_counts.max()
print("Shows the area with the highest number of cases.")
print("Area with the most paper leak cases:", highest_area)
print("Number of cases:", highest_area_cases)
```

# 3. Which exam had the most paper leak cases?

```python
# Exam with the Most Paper Leak Cases

exam_counts = df['exam_name'].value_counts()

highest_exam = exam_counts.idxmax()
highest_exam_cases = exam_counts.max()
print("Shows the exam affected by the most cases.")
print("Exam with the most paper leak cases:", highest_exam)
print("Number of cases:", highest_exam_cases)
```

# 4. What is the most common leak status?

```python
# Most Common Leak Status

status_counts = df['leak_status'].value_counts()

common_status = status_counts.idxmax()
common_status_cases = status_counts.max()
print("Shows the most common leak status.")
print("Most common leak status:", common_status)
print("Number of cases:", common_status_cases)
```

# 5. Which era recorded more paper leak cases?

```python
# Compare UPA and NDA Eras

era_counts = df['era'].value_counts()

for era, count in era_counts.items():
    print(era, ":", count, "cases")
    print("Shows which era has more cases in our dataset.")
```

# 6. What does the relationship between arrests and convictions show?

```python
# Arrests and Convictions Relationship

correlation = df['arrests'].corr(df['convictions'])
print("Shows the relationship between arrests and convictions.")
print("Correlation between arrests and convictions:", round(correlation, 2))
```
