# 5. Selecting Required Columns

```python
selected_data = df[
    [
        "incident_id",
        "date",
        "era",
        "exam_name",
        "conducting_body",
        "body_type",
        "area",
        "leak_status",
        "action_taken",
        "arrests",
        "convictions",
        "aspirants_affected",
        "confidence"
    ]
]
display(selected_data.head())
```

# 6. Filtering Confirmed Paper Leak Cases

```python
confirmed_leaks = df[df["leak_status"] == "Confirmed"]
display(confirmed_leaks)
print("Number of confirmed leaks: ",confirmed_leaks.shape)
```

# 7. Filtering Alleged Paper Leak Cases

```python
alleged_cases = df[df["leak_status"] == "Alleged"]

display(alleged_cases)
print("Shape of Alleged Cases:", alleged_cases.shape)
```

# 8. Filtering Denied Paper Leak Cases

```python
denied_cases = df[df["leak_status"] == "Denied"]

display(denied_cases)
print("Shape of Denied Cases:", denied_cases.shape)
```

# 9. Filtering Suspected Paper Leak Cases

```python
suspected_cases = df[df["leak_status"] == "Suspected"]

display(suspected_cases)
print("Shape of Suspected Cases:", suspected_cases.shape)
```

# 10. Filtering Central Conducting Body Cases

```python
central_cases = df[df["body_type"] == "Central"]

display(central_cases)

print("Number of central_cases : ",central_cases.shape)
```

#11.  Filtering State Conducting Body Cases

```python
state_cases = df[df["body_type"] == "State"]

display(state_cases)

print("Number of state_cases : ",state_cases.shape)
```

# 12. Filtering NDA Era Paper Leak Cases

```python
nda_cases = df[df["era"] == "NDA (May2014-now)"]
display(nda_cases)
print("Number of nda_cases : ",nda_cases.shape)
```

# 13. Filtering UPA Era Paper Leak Cases

```python
upa_cases = df[df["era"] == "UPA (2004-May2014)"]

display(upa_cases)

print("Number of UPA cases:", upa_cases.shape)
```

# 14. Filtering High Confidence Paper Leak Cases

```python
high_confidence = df[df["confidence"] == "High"]

display(high_confidence)
print("Shape of High Confidence Cases:", high_confidence.shape)
```

# 15. Filtering Medium Confidence Paper Leak Cases

```python
medium_confidence = df[df["confidence"] == "Medium"]

display(medium_confidence)
print("Shape of Medium Confidence Cases:", medium_confidence.shape)
```

# 16. Filtering Low Confidence Paper Leak Cases

```python
low_confidence = df[df["confidence"] == "Low"]

display(low_confidence)
print("Shape of Low Confidence Cases:", low_confidence.shape)
```

# **Filtering Cases with Arrest Information**

```python
arrest_cases = df[df["arrests"].notna()]

display(arrest_cases)
print("Shape of Arrest Cases:", arrest_cases.shape)
```

# **Filtering Cases with Conviction Information**

```python
conviction_cases = df[df["convictions"].notna()]

display(conviction_cases)
print("Shape of Conviction Cases:", conviction_cases.shape)
```
